#**PROBABILIDAD Y ESTADISTICA**
##***La estadistica***
 ### Es una rama de la organizacion y realizacion de los dtaos. se usan para comprender tendencias, relaciones y conjuntos de informacio. el objetivo de la estadistica sirve como una herramienta de guia para un respaldo de dato, se pueden tomar decisiones y tener menor indice de fallo

##***Definicion de logica:***
### Es un metodo de razonamiento que combina la logica deductiva con la teoria de probabilidad. su principal objetivo es tratar situaciones inciertas.

##***Estadistica discriptiva e inferencial***
###-Sirve en economia, inversiones, medicina
###-Probabilidad rama de la matematica que estudia o investiga la probabilidad de los eventos
###-La estadistica es una herramienta esencial para persivir cosas del mundo que nos rodea, es necesario ver un conjunto de los eventos posibles.

##***Estadistica descriptiva***
###-Resumir caracteristicas principales de un conjunto de datos organizar e interpretar diferente informacion
###-Se centra en describir las propiedades del producto deseado.
###-Se puede desglozar por sus pasos principales, recoleccion de datos relevantes, se organizar en tablas o graficos para facilitar el analisis.
###-Medidas central, medidas de formas cimetricas, identificacion de patrones y variaciones
###-Media, mediana, medidas de posicionamiento. dispercion
###-Graficas pictograma, poligono de frecuencias, tallos y hojas, caja
###-Es una herramienta donde se puede ajustar la informacion para que se pueda entender con mas facilidad.

##***Estadistica inferencial***
###-Se encarga de reunir informacion de una poblacion apartir de una muestra. principio de intercidumbre.
###-busca realizar predicciones y tomar decisiones en una poblacion mas amplea.
###-Estigmaciones se basan por la muestra, se generan pruebas de hipoticis, intervalos de confianza, modelado de estadistica aleaterio, sistematico. pruebas de hipotezis, vectores, etc.
###-Estigmacion puntual.
###-Conceptos claves, probabilidad, distribucion, probabilidad.
###-Se usa para estudios medicos, economicos, investigacion cientifica, marketing, finanzas, ciencias sociales.
###-Comprende los fenomenos que ocurren en el mundo real.

##**Ejemplo estadistico**
##***Ejemplo analisis de rendimiento de sistema de streaming***
###Se recopilaron datos durante 2 días y 2 horas acerca de los servidores de una plataforma de streaming. A partir de estos datos, podemos formular  hipótesis  entre las 5:00 p.m. y las 8:00 p.m., las respuestas de los servidores aumentan considerablemente. Esto podría deberse a las posibles razones.

###***Primera hipótesis:*** El incremento registrado dentro de la medición pudo haberse provocado por un resultado de una mala medición o de alguna interferencia generada ajena al servidor

###***Segunda hipótesis:*** La latencia generada pudo ser provocada por uno o varios eventos transmitidos de manera simultánea, lo que trae como consecuencia una sobrecarga mediante la alta demanda del servicio de streaming.

###***Tercera hipótesis:*** La infraestructura del servidor no esta lo suficientemente suministrada para soporta un alta demanda de los usuarios, lo que se puede deducir como bases de datos lentas y/o filas largas de procesamiento por parte del servidor.

###***Generalizando:***  El incremento de la latencia registrado en la medición no necesariamente es el mismo a la misma hora todos los días, ya que la medición resulta ser solo un amuestra de todos los eventos posibles que pueden generar este incremento en la velocidad de respuesta.

###Para concluir con las deducciones establecidas, l a empresa decidió mejorar las consultas a la base de datos. Con esto habría una mejora en todo el servidor, en cuanto a las recepciones, habrían incluso más visitas a la página por una mejor optimización durante las horas pico.

###Con esta mejora se espera que el tiempo se reduzca, incluso que sea mejor que el tiempo promedio que es de 2.4 segundos en las horas menos activas. Esto le daría una mejor experiencia a las personas al momento de usar el servidor. Se ofrecería carga de información más rápida y la reducción de espera.
